# Notebook 01: Exploratory Data Analysis & Quality Validation
**Retail Demand Forecasting & Inventory Intelligence Platform**

This notebook inspects the raw datasets (train, stores, oil, holidays, transactions), validates schema consistency, and evaluates weekly seasonality, promotional lifts, and store sales distributions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Ensure project root in path
import sys
sys.path.append('..')
from src.config import settings
from src.data.loader import load_raw_data, validate_raw_data

# Load datasets
dfs = load_raw_data(auto_generate=True)
train = dfs['train']
stores = dfs['stores']
oil = dfs['oil']
holidays = dfs['holidays']
transactions = dfs['transactions']

print("Train shape:", train.shape)
train.head()
```

### Step 1: Data Quality Assertions

```python
checks = validate_raw_data(dfs)
for check, passed in checks.items():
    print(f"Assertion '{check}': {'PASSED' if passed else 'FAILED'}")
```

### Step 2: Sales Distribution Across Product Families

```python
plt.figure(figsize=(10, 4))
family_volume = train.groupby('family')['sales'].sum().sort_values(ascending=False)
sns.barplot(x=family_volume.values, y=family_volume.index, palette='Blues_r')
plt.title('Total Demand Volume by Product Category')
plt.xlabel('Cumulative Unit Sales')
plt.show()
```

### Step 3: Day-of-Week Seasonality

```python
train['date'] = pd.to_datetime(train['date'])
train['day_of_week'] = train['date'].dt.day_name()
dow_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

plt.figure(figsize=(8, 4))
sns.barplot(data=train, x='day_of_week', y='sales', order=dow_order, palette='viridis')
plt.title('Average Daily Sales by Day of Week')
plt.ylabel('Mean Units Sold')
plt.xticks(rotation=45)
plt.show()
```
